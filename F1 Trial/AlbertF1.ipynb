```python
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import pandas as pd
import fastf1
from fastf1 import get_session, plotting
from datetime import timedelta
from typing import Tuple, Union

# Setup Fast F1 cache and matplotlib configurations
fastf1.Cache.enable_cache(r"C:\Users\moses_y\OneDrive\Desktop\ML Projects\F1 Trial")
plotting.setup_mpl()

class F1DataAnalysis:
    def __init__(self, year: int, grand_prix: str):
        self.year = year
        self.grand_prix = grand_prix
        self.quali_session = get_session(year, grand_prix, 'Q')
        self.race_session = get_session(year, grand_prix, 'R')

    @staticmethod
    def format_timedelta(td: timedelta) -> str:
        """ Converts a pandas Timedelta object to a formatted string. """
        minutes = int(td.total_seconds() // 60)
        seconds = td.total_seconds() % 60
        return f"{minutes:02d}:{seconds:05.2f}"

    def load_sessions(self):
        """ Loads and caches the qualifying and race session data. """
        self.quali_session.load()
        self.race_session.load()

    def plot_fastest_laps(self, drivers: list, use_plotly: bool = True) -> None:
        """ Plots the fastest laps using either Plotly or Matplotlib based on the flag. """
        if use_plotly:
            fig = go.Figure()
        else:
            plt.figure(figsize=(20, 10))

        for session, session_name in [(self.quali_session, 'Qualifying'), (self.race_session, 'Race')]:
            for driver in drivers:
                laps = session.laps.pick_driver(driver)
                fastest_lap = laps.pick_fastest()
                if fastest_lap.empty:
                    print(f"Warning: No data for {driver} during {session_name}.")
                    continue

                if 'Time' in fastest_lap.columns and 'Speed' in fastest_lap.columns:
                    times = fastest_lap['Time'].dt.total_seconds()
                    speeds = fastest_lap['Speed']
                    lap_time = self.format_timedelta(fastest_lap['LapTime'].iloc[0])
                    if use_plotly:
                        fig.add_trace(go.Scatter(x=times, y=speeds, mode='lines', name=f'{driver} {session_name} Fastest Lap ({lap_time})'))
                    else:
                        plt.plot(times, speeds, label=f'{driver} {session_name} Fastest Lap ({lap_time})')
                else:
                    print(f"Missing data for {driver} in {session_name}")

        if use_plotly:
            fig.update_layout(
                title=f'{self.year} {self.grand_prix} - Fastest Laps',
                xaxis_title='Time (s)', yaxis_title='Speed (km/h)',
                legend_title='Driver & Session', template='plotly_dark')
            fig.show()
        else:
            plt.title(f'{self.year} {self.grand_prix} - Fastest Laps')
            plt.xlabel('Time (s)')
            plt.ylabel('Speed (km/h)')
            plt.legend()
            plt.tight_layout()
            plt.show()

def main():
    analysis = F1DataAnalysis(2024, 'Miami Grand Prix')
    analysis.load_sessions()
    analysis.plot_fastest_laps(['LEC', 'VER'])

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import fastf1
from fastf1 import get_session, plotting
from datetime import datetime, timedelta
import requests
import pandas as pd

# Setup Fast F1
cache_path = r"C:\Users\moses_y\OneDrive\Desktop\ML Projects\F1 Trial"
fastf1.Cache.enable_cache(cache_path)
plotting.setup_mpl()

def format_timedelta(timedelta):
    """ Convert a timedelta object into a formatted string. """
    minutes, seconds = divmod(int(timedelta.total_seconds()), 60)
    return f"{minutes}:{seconds:02d}"

def fetch_f1_schedule(year):
    """ Fetches the F1 schedule for a given year. """
    url = f'https://ergast.com/api/f1/{year}.json'
    try:
        response = requests.get(url)
        races = response.json()['MRData']['RaceTable']['Races']
        return {race['raceName']: race['date'] for race in races}
    except requests.RequestException as e:
        print(f"Error fetching schedule: {e}")
        return {}

def is_race_week(race_date):
    """ Check if the current date is within the race week. """
    race_date = datetime.strptime(race_date, '%Y-%m-%d')
    start_week = race_date - timedelta(days=race_date.weekday())
    end_week = start_week + timedelta(days=6)
    return start_week <= datetime.now() <= end_week

def plot_fastest_laps(year, grand_prix, drivers, use_plotly=True):
    """ Plot the fastest laps using either Plotly or Matplotlib. """
    sessions = ['Q', 'R']  # Example session codes for Qualifying and Race
    try:
        for session_code in sessions:
            session = get_session(year, grand_prix, session_code)
            session.load()
            print(f"Data loaded for {grand_prix} - Session: {session_code}")

            if use_plotly:
                fig = go.Figure()
            else:
                plt.figure(figsize=(20, 10))

            for driver in drivers:
                laps = session.laps.pick_driver(driver)
                fastest_lap = laps.pick_fastest()
                if fastest_lap.empty:
                    print(f"No data for {driver} in {session_code}")
                    continue

                times = [lap['Time'].total_seconds() for lap in fastest_lap]
                speeds = [lap['Speed'] for lap in fastest_lap]
                lap_time = format_timedelta(fastest_lap['LapTime'])

                if use_plotly:
                    fig.add_trace(go.Scatter(x=times, y=speeds, mode='lines', name=f'{driver} - {lap_time}'))
                else:
                    plt.plot(times, speeds, label=f'{driver} - {lap_time}')

            if use_plotly:
                fig.update_layout(title=f'Fastest Laps - {year} {grand_prix}',
                                  xaxis_title='Time (s)', yaxis_title='Speed (km/h)',
                                  legend_title='Driver & Session', template='plotly_dark')
                fig.show()
            else:
                plt.title(f'Fastest Laps - {year} {grand_prix}')
                plt.xlabel('Time (s)')
                plt.ylabel('Speed (km/h)')
                plt.legend()
                plt.show()
    except Exception as e:
        print(f"An error occurred while plotting: {e}")

def plot_f1_season(year):
    """ Check and plot data for races up to today's date within the current season. """
    schedule = fetch_f1_schedule(year)
    for gp, date in schedule.items():
        if is_race_week(date):
            print(f"Today is race week for {gp}. Plotting...")
            plot_fastest_laps(year, gp, ['LEC', 'VER'], use_plotly=True)

if __name__ == "__main__":
    plot_f1_season(datetime.now().year)
```
