## Exploratory Data Analysis (EDA) phase.

In the EDA phase, we aim to uncover patterns, relationships, and insights from the data. This will involve:

Statistical Summary: Analyzing the central tendencies and variances of the numerical features to understand the data's distribution.
Correlation Analysis: Investigating how the features relate to each other and to the employee performance rating.
Visualization: Creating visual representations of the data to identify patterns, trends, and outliers.
Let's begin by providing a statistical summary and performing a correlation analysis to identify the features most related to employee performance.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df1 = pd.read_excel(xls, sheet_name=0) # Assuming 'xls' is previously loaded ExcelFile

# Plotting histograms for numerical features
numerical_features = df1.select_dtypes(include=['int64', 'float64']).columns
df1[numerical_features].hist(bins=15, figsize=(15, 6), layout=(5, 4))
plt.tight_layout()
plt.show()

# Density Plots for a closer look at distributions
for col in numerical_features:
    sns.displot(df1[col], kind="kde", fill=True)
    plt.title(f'Density Plot of {col}')
    plt.xlabel(col)
    plt.ylabel('Density')
    plt.show()
```

```python
# Exploring relationships using scatter plots for a selection of features against PerformanceRating
# Replace 'Feature1', 'Feature2' with actual feature names as needed
for col in numerical_features:
    sns.violinplot(data=df1, x=col, y='PerformanceRating')
    plt.title(f'Scatter Plot of {col} vs. PerformanceRating')
    plt.show()

# Box plots to examine distributions of features by PerformanceRating
for col in numerical_features:
    sns.boxplot(data=df1, x='PerformanceRating', y=col)
    plt.title(f'Box Plot of {col} by PerformanceRating')
    plt.show()
```

```python
# Correlation Analysis to identify potential predictors of employee performance
correlation_matrix = df1.corr()

plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Features')
plt.show()

# Noting features with strong correlations for potential feature engineering
strong_correlated_features = correlation_matrix['PerformanceRating'][abs(correlation_matrix['PerformanceRating']) > 0.1].index.tolist()
strong_correlated_features.remove('PerformanceRating')  # Removing the target variable itself

# For demonstration, let's create new features based on the insights from EDA
# Creating an interaction term between 'Age' and 'TotalWorkExperienceInYears' as they may have a combined effect
df1['Age_WorkExp_Interaction'] = df1['Age'] * df1['TotalWorkExperienceInYears']

# Recalculate the correlation matrix including the new feature
correlation_matrix_new = df1[['Age_WorkExp_Interaction', 'PerformanceRating']].corr()

# Visualizing the updated correlation matrix with the new feature
sns.heatmap(correlation_matrix_new, annot=True, cmap='viridis', fmt='.2f')
plt.title('Correlation Matrix with New Feature')
plt.show()
```

## The Exploratory Data Analysis (EDA) reveals several insights:

### Statistical Summary Highlights:
The dataset spans employees aged 18 to 60, with an average age of approximately 37 years.
The EmpEnvironmentSatisfaction, EmpLastSalaryHikePercent, and EmpWorkLifeBalance have varying degrees of correlation with PerformanceRating, suggesting these factors might significantly impact employee performance.

### Correlation Analysis:
Positive Correlations: EmpEnvironmentSatisfaction and EmpLastSalaryHikePercent show positive correlations with PerformanceRating, indicating that higher satisfaction with the work environment and larger salary hikes are associated with higher performance ratings.
Negative Correlations: Features like YearsSinceLastPromotion, EmpDepartment, and ExperienceYearsInCurrentRole are negatively correlated with performance ratings, suggesting that longer periods without promotion and certain departmental roles might be linked to lower performance ratings.

### Visualization Insights:
The heatmap illustrates the relationships between variables, highlighting how some features are more strongly correlated with employee performance than others.

## Feature Engineering:
Given these insights, we might consider:

Creating Interaction Features: For variables that are notably correlated with performance ratings, such as `EmpEnvironmentSatisfaction` and `EmpLastSalaryHikePercent`.
Aggregating Features: For example, combining `YearsSinceLastPromotion`, `ExperienceYearsAtThisCompany`, and `ExperienceYearsInCurrentRole` into a single metric that reflects overall career progression within the company.

This step will involve creating new features based on our EDA findings and preparing the data for machine learning models.

```python
# Feature Engineering: Creating new features based on EDA insights

# Interaction between environment satisfaction and last salary hike percent
df1['EnvSatisfaction_SalaryHike'] = df1['EmpEnvironmentSatisfaction'] * df1['EmpLastSalaryHikePercent']

# Aggregating years since last promotion, experience years at this company, and experience years in current role
df1['CareerProgression'] = df1['YearsSinceLastPromotion'] + df1['ExperienceYearsAtThisCompany'] + df1['ExperienceYearsInCurrentRole']

# Re-checking the correlation of the new features with PerformanceRating
new_feature_correlation = df1[['EnvSatisfaction_SalaryHike', 'CareerProgression', 'PerformanceRating']].corr()['PerformanceRating']

new_feature_correlation
```



The new features created during the feature engineering process show the following correlations with `PerformanceRating`:

- **`EnvSatisfaction_SalaryHike` (0.533085):** This feature, an interaction between `EmpEnvironmentSatisfaction` and `EmpLastSalaryHikePercent`, shows a strong positive correlation with `PerformanceRating`. This suggests that a combination of high environment satisfaction and significant salary hikes might be a strong predictor of high employee performance.
- **`CareerProgression` (-0.152728):** An aggregate feature combining `YearsSinceLastPromotion`, `ExperienceYearsAtThisCompany`, and `ExperienceYearsInCurrentRole`, has a negative correlation with performance ratings. This indicates that longer career progression within the company might be associated with lower performance ratings, potentially reflecting a plateau effect where longer tenure in roles without change might not necessarily lead to higher performance ratings.

These insights support the idea that not only standalone features but also their interactions and aggregate measures can provide valuable signals about employee performance.
