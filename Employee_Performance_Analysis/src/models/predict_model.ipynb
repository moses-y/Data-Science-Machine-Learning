### Steps to Implement in `predict_model.ipynb`:

1. **Load the Best Model:**
   Load the Gradient Boosting model, which showed the highest performance in your analysis. If you've saved the model using `joblib`, you can load it as follows:


```python
import pandas as pd
import joblib
best_model = joblib.load('Gradient Boosting.joblib')

```


2. **Prepare New/Unseen Data:**
   Ensure any new data for prediction is preprocessed in the same way as the training data. This includes scaling numerical features and applying one-hot encoding to categorical features.

3. **Make Predictions:**
   Use the loaded model to predict the performance ratings of new or unseen employees.
   

```python
processed_file_path = r'C:\Users\moses_y\OneDrive\Desktop\ML Projects\Employee_Performance_Analysis\data\processed\INX_Future_Inc_Employee_Performance_Preprocessed.xlsx'
data = pd.read_excel(processed_file_path)

X = data.drop('PerformanceRating', axis=1)

predictions = best_model.predict(X)  # Assuming X_new is your preprocessed new data

print(predictions)
```


4. **Interpret Predictions:**
   Analyze the prediction results. You can categorize employees based on their predicted performance ratings to identify high performers, average performers, and those who may need support or development.

```python
performance_labels = {-1: 'Error', 0: 'Low', 1: 'Medium', 2: 'High'}
predicted_labels = [performance_labels[pred] for pred in predictions]

print(predicted_labels[:5])
```
