```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from scipy.sparse import csr_matrix

# Load the dataset
file_path = r'C:\Users\moses_y\OneDrive\Desktop\ML Projects\Employee_Performance_Analysis\data\raw\INX_Future_Inc_Employee_Performance_CDS_Project2_Data_V1.8.xlsx'
data = pd.read_excel(file_path)

# Setting EmpNumber as the index
data.set_index('EmpNumber', inplace=True)

# Define numerical and categorical columns
numerical_cols = data.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = data.select_dtypes(include=['object']).columns

# Define the Pipeline steps for numerical and categorical data preprocessing
numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combine transformers into a ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ]
)

# Apply the ColumnTransformer to the dataset
data_preprocessed = preprocessor.fit_transform(data)

# Convert the preprocessed data back to a DataFrame
# If the output is a sparse matrix, convert it to a dense format
if isinstance(data_preprocessed, csr_matrix):
    data_preprocessed = data_preprocessed.toarray()
# Get feature names for transformed categorical columns
categorical_features = preprocessor.named_transformers_['cat']['onehot'].get_feature_names_out(categorical_cols)
all_features = np.append(numerical_cols, categorical_features)

data_preprocessed_df = pd.DataFrame(data_preprocessed, columns=all_features)

# Save the preprocessed data to the processed folder
processed_file_path = r'C:\Users\moses_y\OneDrive\Desktop\ML Projects\Employee_Performance_Analysis\data\processed\INX_Future_Inc_Employee_Performance_Preprocessed.xlsx'
data_preprocessed_df.to_excel(processed_file_path, index=False)

print(f"Preprocessed data has been successfully saved to {processed_file_path}.")
```
