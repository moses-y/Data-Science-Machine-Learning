```python
# Analyzing Average Spending by Age Group.
def analyze_age(df):
    age_groups = pd.cut(df['Age'], bins=[0, 18, 30, 45, 60, 100], labels=['0-18', '19-30', '31-45', '46-60', '60+'])
    df['Age Group'] = age_groups

    age_vs_spend = df.groupby('Age Group')['Total Spend'].mean()
    px.bar(x=age_vs_spend.index, y=age_vs_spend.values, labels={'x':'Age Group', 'y':'Average Spend'}, title='Average Spending by Age Group').show()

analyze_age(df)
```

```python
# Visualizing a 3D scatter plot of Age, Total Spend, and Satisfaction Level.
def scatter_3d(df):
    fig = px.scatter_3d(df, x='Age', y='Total Spend', z='Satisfaction Level', color='Satisfaction Level', size='Total Spend')
    fig.update_layout(scene=dict(xaxis_title='Age', yaxis_title='Total Spend', zaxis_title='Satisfaction Level'))
    fig.show()

scatter_3d(df)
```

```python
# Visualizing spending based on Membership Type and whether a discount was applied.
def spending_by_membership_and_discount_plotly(df):
    # Grouping and unstacking data.
    df_grouped = df.groupby(['Membership Type', 'Discount Applied'])['Total Spend'].sum().unstack().reset_index()
    
    # Creating a stacked bar chart.
    fig = px.bar(df_grouped, x='Membership Type', y=df_grouped.columns[1:], 
                 title='Spending by Membership Type and Discount Applied',
                 labels={'value': 'Total Spend', 'variable': 'Discount Applied'},
                 barmode='stack')

    fig.update_layout(
        xaxis_title='Membership Type',
        yaxis_title='Total Spend'
    )
    
    fig.show()

spending_by_membership_and_discount_plotly(df)
```

```python
# Performing clustering and visualizing results.
def clustering_with_plotly(df, features, n_clusters, cluster_name):
    scaler = StandardScaler()
    scaled_features = scaler.fit_transform(df[features])

    model = KMeans(n_clusters=n_clusters, random_state=42)
    df[cluster_name] = model.fit_predict(scaled_features)

    cluster_distribution = df[cluster_name].value_counts().sort_index()
    
    # Creating bar chart.
    fig = go.Figure(data=[
        go.Bar(x=cluster_distribution.index, y=cluster_distribution.values)
    ])
    fig.update_layout(
        title=f'Customer Distribution Across {cluster_name}',
        xaxis_title=cluster_name,
        yaxis_title='Number of Customers'
    )
    fig.show()

clustering_with_plotly(df, features=['Age', 'Total Spend', 'Items Purchased', 'Average Rating'], n_clusters=5, cluster_name='Customer Segment')
```
