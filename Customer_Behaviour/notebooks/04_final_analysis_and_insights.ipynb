PURPOSE & APPROACH
• We analyzed 49,661 unique customers using two segmentation techniques:
– RFM (Recency, Frequency, Monetary) to classify customers into groups like “Champions,” “Lost Champions,” etc.
– Customer Lifetime Value (CLTV) to categorize them into “Bronze,” “Silver,” “Gold,” and “Platinum.”
• These combined methods reveal both short-term purchasing behavior (RFM) and long-term revenue potential (CLTV).


KEY OBSERVATIONS
a) RFM Segments
• Average RFM metrics across all customers:
– Recency ≈ 743 days, Frequency ≈ 5 orders, Monetary ≈ $3,856.
• Largest segment is “Lost Low Value” with 11,221 customers—long time since last purchase, low frequency, lower spend.
• “Champions” (8,110 customers) are high-value regular buyers with recent purchases. They represent prime candidates for loyalty programs.


b) CLTV Segments
• Mean CLTV: ≈ $1,178, ranging from as low as ~$0.01 to nearly $18,417.
• “Platinum” customers hold the top quartile of CLTV; they typically show higher Average Order Value and purchase frequency.
• “Bronze” segment has smaller transaction counts/prices, translating into the lowest CLTV.


STRATEGIC INSIGHTS
• Retention vs. Acquisition:
– “Champions” (RFM) and “Platinum” (CLTV) should receive retention-focused investments (premium loyalty perks, personalized offers).
– “Lost Low Value” and “Bronze” require cost-effective win-back or reactivation strategies, either via discounts or targeted campaigns.
• Segment Overlap:
– Not all “Champions” are “Platinum”; some fall into “Gold.” Upselling or bundling may help lift them into higher-value segments.
– “Lost Champions” (3,396 customers) frequently overlap with “Gold” or “Platinum” tiers, indicating they were once heavy spenders but are now drifting away—an ideal group for re-engagement.
• Marketing Budget Allocation:
– High potential ROI when focusing on retaining or upselling top segments (“Champions,” “Platinum,” “Lost Champions”).
– Lower-tier segments may benefit from broad-based or automated marketing to reduce churn cost.


NEXT STEPS
• Investigate Churn Triggers: Explore factors behind “Lost Customers” and “Lost Low Value” to address churn causes (return policies, pricing, shipping issues).
• Micro-Segmentation: Combine RFM+CLTV with demographic or behavioral data (gender, geography) for more finely tuned campaigns.
• Integration with Churn Modeling: Use RFM and CLTV features to improve predictive accuracy of churn models and allocate marketing resources effectively.

TITLE: Churn Modeling for Enhanced Customer Retention


DATA & CONTEXT
• Volume & Scope: 250,000 records, ~38 columns tracking customer demographics, purchase history, returns, RFM/CLTV metrics, and a “Churn” indicator.
• Purpose: Identify factors that drive churn (20% of customers), enabling targeted interventions to boost retention.


KEY INSIGHTS & EDA FINDINGS
• Overall Churn Rate: ~20% (1 in 5 customers).
• Payment & Churn:
– Cash (20.25%) & PayPal (20.24%) churn slightly higher than Credit Card (19.66%). This suggests automated or recurring billing fosters loyalty.
• Product Category & Churn:
– Clothing leads (20.12%) vs. Electronics (20.04%) & Home (19.96%). Investigate product fit or return policies for possible improvements.
• Gender & Churn:
– Female: 20.16% vs. Male: 19.94%. Gap is small, so retention approaches can be similar for both.
• Value Segments (CLTV) & Churn:
– “Gold” customers show the highest churn (20.42%), indicating high-value clients with potentially higher service expectations.
• Correlation Matrix:
– Churn has moderate to low correlation with single features, suggesting churn is multifactorial. Frequency, Monetary, Total Price/Revenue, and CLTV strongly correlate with one another (≥0.70), reflecting that frequent/high-spend shoppers have high lifetime value.


MODELING APPROACH & RESULTS
• Classifiers Used: Logistic Regression, Random Forest, XGBoost, plus a simple Neural Network.
• Imbalanced Data Handling: SMOTE oversampling to balance the ~20% churn minority class.
• Performance Highlights:
– Random Forest: ~81% Accuracy, ~32% F1-score for churn class, best overall recall for the positive (churn) class (22%).
– Neural Network: ~78% Accuracy, though struggled with recall (3% for churn). Model can be further tuned (more epochs, different architecture).
• Feature Importance (Random Forest): Recency, Frequency, and Monetary (RFM) metrics dominate, echoing the importance of purchase behavior in predicting churn.


BUSINESS RELEVANCE & NEXT STEPS
• Tailored Retention: Focus on gold-tier or higher spenders with high risk (low recency, high monetary). Offer loyalty perks, special promotions, or automated payment incentives.
• Product & Payment Methods: Clothing category and one-off Cash/PayPal buyers are slightly more prone to churn; consider subscription billing or targeted re-engagement.
• Model Refinements:
– Hyperparameter tuning (GridSearchCV) on top models (RF, XGB).
– Examine advanced interpretability with SHAP/LIME.
– Integrate triggers (e.g., days since last purchase) to proactively flag at-risk customers.
