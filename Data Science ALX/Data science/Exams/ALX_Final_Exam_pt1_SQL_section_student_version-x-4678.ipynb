```python
import pandas as pd
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import SnowballStemmer
from sklearn.feature_extraction.text import CountVectorizer

# Download NLTK resources
nltk.download('stopwords')
nltk.download('punkt')

# Load the Essay data
essay_data_df = pd.read_csv(r"C:\Users\moses_y\OneDrive\Desktop\ML Projects\alx\Data science\Exams\Exam_data_part_2-20240731T160327Z-001\Exam_data_part_2\Essay_data.csv")

# Remove rows with missing values and reset the index
essay_data_df = essay_data_df.dropna().reset_index(drop=True)

# 42. Number of bi-grams after text preprocessing
sentence = "I’m a part-time student @explore-software."

# Preprocessing steps
sentence = sentence.translate(str.maketrans(string.punctuation, ' ' * len(string.punctuation)))
sentence = sentence.lower()
stop_words = set(stopwords.words('english'))
tokens = [word for word in word_tokenize(sentence) if word.isalnum() and word not in stop_words]
bigrams = list(nltk.bigrams(tokens))
print(f"42. Number of bi-grams: {len(bigrams)}")

# 43. Ratio of intuitive (N) students to sensing (S) students
intuitive_count = essay_data_df[essay_data_df['N/S'].str.contains('N')].shape[0]
sensing_count = essay_data_df[essay_data_df['N/S'].str.contains('S')].shape[0]
print(f"43. Ratio of intuitive to sensing students: {intuitive_count}:{sensing_count}")

# 44. 10th character in the first essay
first_essay = essay_data_df['Essay'][0]
first_essay_clean = first_essay.translate(str.maketrans('', '', string.punctuation)).lower()
tenth_character = first_essay_clean[9]
print(f"44. 10th character in the first essay: '{tenth_character}'")

# 45. Number of tokens in the 17th essay
seventeenth_essay = essay_data_df['Essay'][16]
tokens_17th = word_tokenize(seventeenth_essay)
tokens_count_17th = len(tokens_17th)
print(f"45. Number of tokens in the 17th essay: {tokens_count_17th}")

# 47. SnowballStemmer output for "experiences"
stemmer = SnowballStemmer('english')
stemmed_word = stemmer.stem('experiences')
print(f"47. SnowballStemmer output for 'experiences': '{stemmed_word}'")

# 48. 24th token in the 81st essay after removing stopwords
eighty_first_essay = essay_data_df['Essay'][80]
tokens_81st = [word for word in word_tokenize(eighty_first_essay) if word.isalnum() and word.lower() not in stop_words]
token_24th_81st = tokens_81st[23]
print(f"48. 24th token in the 81st essay: '{token_24th_81st}'")

# 49. Number of unique words after removing stopwords
all_words = [word.lower() for text in essay_data_df['Essay'] for word in word_tokenize(text) if word.isalnum() and word.lower() not in stop_words]
unique_words_count = len(set(all_words))
print(f"49. Number of unique words: {unique_words_count}")

# 50. Number of times 'time' was mentioned in the 56th essay using bag of words
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(essay_data_df['Essay'])
words = vectorizer.get_feature_names_out()
essay_56 = X[55].toarray()
time_count = essay_56[0, words == 'time']
print(f"50. Number of times 'time' was mentioned in the 56th essay: {time_count[0]}")

# 51. Words that appear at least twice percentage
word_counts = pd.Series(all_words).value_counts()
total_words = len(all_words)
words_at_least_twice = word_counts[word_counts >= 2].sum()
percentage_words_twice = (words_at_least_twice / total_words) * 100
print(f"51. Words that appear at least twice percentage: {percentage_words_twice:.2f}%")

# 52. Most commonly mentioned word by ENFJ personalities
enfj_essays = essay_data_df[(essay_data_df['I/E'] == 'E') & (essay_data_df['N/S'] == 'N') & (essay_data_df['T/F'] == 'F') & (essay_data_df['J/P'] == 'J')]['Essay']
all_enfj_words = [word.lower() for text in enfj_essays for word in word_tokenize(text) if word.isalnum() and word.lower() not in stop_words]
most_common_enfj_word = pd.Series(all_enfj_words).value_counts().idxmax()
print(f"52. Most commonly mentioned word by ENFJ personalities: '{most_common_enfj_word}'")

# 53. 109th bi-gram in the 70th essay
seventieth_essay = essay_data_df['Essay'][69]
tokens_70th = [word.lower() for word in word_tokenize(seventieth_essay) if word.isalnum() and word.lower() not in stop_words]
bigrams_70th = list(nltk.bigrams(tokens_70th))
bigram_109th_70th = bigrams_70th[108] if len(bigrams_70th) > 108 else None
print(f"53. 109th bi-gram in the 70th essay: {bigram_109th_70th}")

results = {
    '42. Number of bi-grams': len(bigrams),
    '43. Ratio of intuitive to sensing students': f"{intuitive_count}:{sensing_count}",
    '44. 10th character in the first essay': tenth_character,
    '45. Number of tokens in the 17th essay': tokens_count_17th,
    '47. SnowballStemmer output for "experiences"': stemmed_word,
    '48. 24th token in the 81st essay': token_24th_81st,
    '49. Number of unique words': unique_words_count,
    '50. Number of times "time" was mentioned in the 56th essay': time_count[0],
    '51. Words that appear at least twice percentage': percentage_words_twice,
    '52. Most commonly mentioned word by ENFJ personalities': most_common_enfj_word,
    '53. 109th bi-gram in the 70th essay': bigram_109th_70th
}

results
```
